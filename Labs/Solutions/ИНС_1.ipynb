1. Построим графики активационных функций: МакКаллока-Питтса, линейную, сигмоидальную и тангенциальную сигмоидальную

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
```

```python
# Диапазон входных значений
x = np.linspace(-6, 6, 1000)

# Пороговая функция Мак-Каллока-Питтса
y_threshold = np.sign(x)

# Линейная функция
y_linear = x

# Сигмоидальная функция
y_sigmoid = 1 / (1 + np.exp(-x))

# Тангенциальная сигмоидальная функция
y_sigmoid_tanh = 2 / (1 + np.exp(-x)) - 1

# Построение графиков
plt.figure(figsize=(8, 8))

plt.plot(x, y_threshold, 'r', label='Мак-Каллок–Питтс')
plt.plot(x, y_linear, 'k', label='Линейная')
plt.plot(x, y_sigmoid, 'b', label='Сигмоидальная')
plt.plot(x, y_sigmoid_tanh, 'g', label='Тангенциальная сигмоидальная')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Активационные функции')
plt.legend()
plt.grid(True)
plt.show()
```

На полученном графике видно, что функции имеют существенно различающийся характер.

* Пороговая функция имеет скачкообразный характер.
* Линейная функция изменяется равномерно.
* Сигмоидальная функция имеет S-образную форму и изменяется в диапазоне от 0 до 1.
* Тангенциальная сигмоидальная функция также имеет S-образную форму, но изменяется от −1 до 1.

2. Для анализа скорости изменения функций рассмотрим их производные.

Производная показывает, насколько сильно изменяется выход функции при изменении входного значения.

Для рассматриваемых функций:

Пороговая функция:

$$ f'(x)=0 $$

практически во всех точках, кроме точки скачка, где производная не определена.

Линейная функция:

$$ f'(x)=1 $$

Сигмоидальная функция:

$$ f'(x)=\frac{e^{-x}}{(1+e^{-x})^2} $$

Тангенциальная сигмоидальная функция:

$$ f'(x)=\frac{2e^{-x}}{(1+e^{-x})^2} $$


```python
# Производная пороговой функции
d_threshold = np.zeros_like(x)

# Производная линейной функции
d_linear = np.ones_like(x)

# Производная сигмоиды
d_sigmoid = np.exp(-x) / (1 + np.exp(-x))**2

# Производная тангенциальной сигмоиды
d_sigmoid_tanh = 2 * np.exp(-x) / (1 + np.exp(-x))**2

# Построение графиков производных
plt.figure(figsize=(10, 6))

plt.plot(x, d_threshold, 'r', label='Мак-Каллок–Питтс')
plt.plot(x, d_linear, 'k', label='Линейная')
plt.plot(x, d_sigmoid, 'b', label='Сигмоидальная')
plt.plot(x, d_sigmoid_tanh, 'g', label='Тангенциальная сигмоидальная')

plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title('Производные активационных функций')
plt.legend()
plt.grid(True)
plt.show()
```

На графике производных видно, что для линейной функции производная постоянна и равна 1. Для пороговой функции производная практически везде равна нулю.

У сигмоидальных функций производная имеет максимальное значение около нуля, а при удалении от нуля в положительную или отрицательную сторону постепенно приближается к нулю.

3. Сравнительный анализ активационных функций

Для сравнения активационных функций рассмотрим диапазон выходных значений, характер изменения, производную и чувствительность к изменению входного сигнала.

  **Пороговая функция Мак-Каллока–Питтса**

Пороговая функция принимает значения −1 или 1. При изменении входного значения выход функции остаётся неизменным до тех пор, пока вход не пересечёт пороговое значение. После пересечения порога происходит резкий скачок выходного сигнала.

Производная функции практически во всех точках равна нулю, а в точке скачка не определена. Поэтому такая функция хорошо подходит для моделирования простого бинарного решения, однако её трудно использовать в методах обучения, основанных на вычислении производных.

  **Линейная функция**

Линейная функция изменяется пропорционально входному сигналу. Её выход не ограничен определённым диапазоном. Производная постоянна и равна 1, поэтому одинаковое изменение входного значения вызывает одинаковое изменение выходного.

Преимуществом линейной функции является отсутствие насыщения. Однако использование только линейной функции в нейронной сети ограничивает её возможности, поскольку последовательное применение линейных преобразований остаётся линейным преобразованием.

  **Сигмоидальная функция**

Сигмоидальная функция имеет S-образную форму и принимает значения от 0 до 1. В области около нуля функция изменяется наиболее быстро. При больших положительных и отрицательных значениях входа график становится практически горизонтальным.

Это явление называется насыщением. В области насыщения производная приближается к нулю, поэтому выход нейрона становится мало чувствительным к изменению входного сигнала.

Максимальное значение производной сигмоидальной функции составляет 0,25.

  **Тангенциальная сигмоидальная функция**

Тангенциальная сигмоидальная функция также имеет S-образную форму, но её выход изменяется в диапазоне от −1 до 1. В отличие от обычной сигмоиды, она симметрична относительно начала координат.

Как и обычная сигмоидальная функция, она имеет область насыщения, где производная стремится к нулю. Максимальное значение производной составляет 0,5.

2. Построим нейрон и проведем обучение на двух пространствах по дельта-правилу

```python
def delta_rule_train(X, D, learning_rate=0.1, epochs=20):

    # Начальные веса
    w = np.array([0.1, -0.1], dtype=float)

    # Начальный bias
    b = 0.0

    # История весов
    weight_history = [w.copy()]
    bias_history = [b]

    # История ошибок
    error_history = []

    for epoch in range(epochs):

        total_error = 0

        for i in range(len(X)):

            # Выход линейного нейрона
            y = np.dot(w, X[i]) + b

            # Ошибка
            error = y - D[i]

            # Изменение весов
            dw = -learning_rate * error * X[i]

            # Изменение bias
            db = -learning_rate * error

            # Обновляем веса
            w = w + dw
            b = b + db

            # Сохраняем траекторию
            weight_history.append(w.copy())
            bias_history.append(b)

            # Квадратичная ошибка
            total_error += error ** 2

        # Средняя ошибка за эпоху
        mean_error = total_error / len(X)
        error_history.append(mean_error)

    return (
        w,
        b,
        np.array(weight_history),
        np.array(bias_history),
        np.array(error_history)
    )
```

```python
w_sep, b_sep, history_sep, bias_history_sep, error_sep = delta_rule_train(
    X_sep,
    D_sep,
    learning_rate=0.01,
    epochs=120
)

print("Итоговые веса:", w_sep)
print("Итоговый bias:", b_sep)
```

## Задание 1. Создать формальную модель нейрона

###Нейрон МакКаллока-Питса

```python
def McCuloh_Pitts_neuron(x=None, w=None, bias=0):
    # x — входные значения
    # w — синаптические веса
    # bias — смещение
    s = np.dot(w, x.T) + bias
    # np.dot — скалярное произведение входов и весов
    y = np.sign(s)
    # sign возвращает:
    # -1, если s < 0
    #  0, если s = 0
    #  1, если s > 0
    return y
```

Проверка на одном объекте

```python
x = np.array([1, 0])
w = np.array([1, -1])

y_out = McCuloh_Pitts_neuron(
    x=x,
    w=w,
    bias=0.3
)

print('ответ нейрона', y_out)
```

Десять случайных объектов

```python
x = np.random.random((10, 2))

w = np.array([1, -1])

y_out1 = McCuloh_Pitts_neuron(
    x=x,
    w=w,
    bias=0.3
)

print('ответ нейрона', y_out1)
```

##Линейный нейрон

```python
def linear_neuron(x=None, w=None, bias=0):
  # x - входы
  # w - веса
  # bias - смещение
  s = np.dot(w, x.T) + bias

  # Линейная активационная функция
  y = s

  return y
```

Проверка

```python
x = np.random.random((10, 2))
w = np.array([1, -1])

y_out1 = linear_neuron(
  x=x,
  w=w,
  bias=0.3
)

print('Ответ нейрона', y_out1)
```

## Сигмоидальный нейрон

```python
def sigmoid_neuron(x=None, w=None, bias=0):
  s = np.dot(w, x.T) + bias
  y = 1 / (1 + np.exp(-s))
  return y
```

Проверка

```python
x = np.random.random((10, 2))
w = np.array([1, -1])

y_out1 = sigmoid_neuron(
  x=x,
  w=w,
  bias=0.3
)

print('Ответ нейрона', y_out1)
```

##Тангенциальный нейрон

```python
def sigmoidt_neuron(x=None, w=None, bias=0):
  s = np.dot(w, x.T) + bias
  y = 2 / (1 + np.exp(-s)) - 1
  return y
```

Проверка

```python
x = np.random.random((10, 2))
w = np.array([1, -1])

y_out1 = sigmoidt_neuron(
  x=x,
  w=w,
  bias=0.3
)

print('Ответ нейрона', y_out1)
```

```python
x = np.linspace(-10, 10, 1000).reshape((1000, 1))
w = np.array([0.9]).reshape((1, 1))
bias = 0.3

y_out1 = McCuloh_Pitts_neuron(x=x, w=w, bias=bias)
y_out2 = sigmoid_neuron(x=x, w=w, bias=bias)
y_out3 = sigmoidt_neuron(x=x, w=w, bias=bias)
y_out4 = linear_neuron(x=x, w=w, bias=bias)


# График 1 — все функции в полном масштабе
plt.figure(figsize=(10, 6))

plt.plot(x[:, 0].T, y_out1[0, :], 'r', label='McCulloch-Pitts')
plt.plot(x[:, 0].T, y_out2[0, :], 'b', label='sigmoid')
plt.plot(x[:, 0].T, y_out3[0, :], 'g', label='sigmoid tanh')
plt.plot(x[:, 0].T, y_out4[0, :], 'k', label='linear')
plt.xlim(-7.5, 7.5)
plt.ylim(-7.5, 7.5)

plt.legend()
plt.title('Ответы нейронов на x — полный масштаб')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()


# График 2 — приближенный масштаб для нелинейных функций
plt.figure(figsize=(10, 6))

plt.plot(x[:, 0].T, y_out1[0, :], 'r', label='McCulloch-Pitts')
plt.plot(x[:, 0].T, y_out2[0, :], 'b', label='sigmoid')
plt.plot(x[:, 0].T, y_out3[0, :], 'g', label='sigmoid tanh')
plt.plot(x[:, 0].T, y_out4[0, :], 'k', label='linear')
plt.ylim(-1.2, 1.2)

plt.legend()
plt.title('Ответы нейронов на x — приближенный масштаб')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```
